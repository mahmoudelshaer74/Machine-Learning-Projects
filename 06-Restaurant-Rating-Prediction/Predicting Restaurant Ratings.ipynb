## Feature Engineering

```python
df.columns
```

```python
X=df[["Average Cost for two","Has Table booking","Has Online delivery","Price range"]]
y=df[["Aggregate rating"]]
```

* Above 2 below 2.5 poor
* Above 2.5 below 3.5 Average
* Above 3.5 below 4 Good
* Above 4 below 4.5 Very Good
* Above 4.5 Excellent

```python
X
```

```python
X["Price range"].value_counts()
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
X["Has Table booking"]=le.fit_transform(X["Has Table booking"])
X["Has Online delivery"]=le.fit_transform(X["Has Online delivery"])
```

```python
X
```

```python
le.classes_
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X=scaler.fit_transform(X)
```

```python
X
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(X,y,test_size=.2)
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error
```

```python
def modelresult(predictions):
    print("Mean Squared Error:{}".format(mean_squared_error (y_test, predictions)))
    print('Root Mean Squared Error:{}'.format(np.sqrt(mean_squared_error(y_test, predictions))))
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
lr_predictions=lr.predict(x_test)
```

```python
modelresult(lr_predictions)
```

```python
from sklearn.svm import SVR
from sklearn.model_selection import GridSearchCV
```

```python
svr_predictions=svr.predict(x_test)
```

```python
from sklearn.tree import DecisionTreeRegressor
DTR=DecisionTreeRegressor()
```

```python
param_grid={"max_depth":[10,20,30],"min_samples_leaf":[1,2,],"min_samples_split":[2,5]}
```

```python
grid_DTR=GridSearchCV(DTR,param_grid)
grid_DTR.fit(x_train,y_train)
```

```python
Tree_predictions=grid_DTR.predict(x_test)
```

```python
modelresult(Tree_predictions)
```

```python
from sklearn.ensemble import RandomForestRegressor
rfr=RandomForestRegressor()
```

```python
param_grid={"max_depth":[10,15,17,20,30],"n_estimators":[2,5,10]}
```

```python
grid_rfr=GridSearchCV(rfr,param_grid)
grid_rfr.fit(x_train,y_train)
```

```python
from sklearn.neighbors import KNeighborsRegressor
KNR=KNeighborsRegressor()
param_grid={"n_neighbors":[2,5,8,10],"weights":['uniform', 'distance'],"algorithm":['auto', 'ball_tree', 'kd_tree', 'brute']}
```

```python
from sklearn.ensemble import AdaBoostRegressor
ada=AdaBoostRegressor()
param_grid={"n_estimators":[50,100,200],"learning_rate":[0.1,0.5,1]}
grid_ada=GridSearchCV(ada,param_grid)
grid_ada.fit(x_train,y_train)
ada_predictions=grid_ada.predict(x_test)
modelresult(ada_predictions)
```
